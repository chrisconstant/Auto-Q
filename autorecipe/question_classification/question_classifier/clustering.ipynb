```python
from sentence_transformers import SentenceTransformer
from sklearn.cluster import KMeans
import pandas as pd
```

```python
model = SentenceTransformer('sentence-transformers/average_word_embeddings_glove.6B.300d')
```

```python
dataset = {}
fnames = ['train', 'val', 'test']
for fname in fnames:
    data = {'label': [], 'question': []}
    with open(f'data/{fname}', 'r') as f:
        for line in f:
            label = int(line.split(' ')[0])
            # throw irrelevant
            if label == 0:
                continue
            question = line[2:]
            data['question'].append(question)
    dataset[fname] = data
```

```python
embeddings = model.encode(dataset['train']['question'])
```

```python
kmeans = KMeans(n_clusters=10).fit(embeddings)
```

```python
kmeans.labels_
```

```python
df = pd.DataFrame({'question': dataset['train']['question'], 'cluster': kmeans.labels_})
```

```python
df.to_csv('data/clusters.csv')
```
