```python
import time, os
from dotenv import load_dotenv
from genai.credentials import Credentials
from genai.model import Model
from genai.schemas import GenerateParams
import numpy as np
```

```python
model_id = 'flan-t5-xl-pt-VQ5QsUX4-2024-01-02-08-18-33'
```

```python
load_dotenv()
api_key = os.getenv("GENAI_KEY", None) 
api_url = os.getenv("GENAI_API", None)
creds = Credentials(api_key, api_endpoint=api_url)
params = GenerateParams(decoding_method="greedy")
model = Model(model_id, params=params, credentials=creds)
```

```python
dataset = {}
fnames = ['train', 'val', 'test']
for fname in fnames:
    data = {'label': [], 'question': []}
    with open(f'data/{fname}', 'r') as f:
        for line in f:
            label = int(line.split(' ')[0])
            question = line[2:]
            data['label'].append(label)
            data['question'].append(question)
    dataset[fname] = data
```

```python
val_preds = model.generate(dataset['val']['question'])
test_preds = model.generate(dataset['test']['question'])
```

```python
val_preds = [pred.generated_text for pred in val_preds]
test_preds = [pred.generated_text for pred in test_preds]
```

```python
val_correct = (np.array(val_preds).astype(int) == np.array(dataset['val']['label']).astype(int)).sum()
val_accuracy = val_correct / len(dataset['val']['label'])
```

```python
test_correct = (np.array(val_preds).astype(int) == np.array(dataset['test']['label']).astype(int)).sum()
test_accuracy = test_correct / len(dataset['test']['label'])
```

```python
print(f'val_accuracy:{val_accuracy}, test_accuracy:{test_accuracy}')
```
