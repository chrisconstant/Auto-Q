```python
import pandas as pd
from evaluate import load
import numpy as np

def get_perplexity(questions):
    perplexity = load("perplexity", module_type="metric")
    results = perplexity.compute(model_id='gpt2',
                                add_start_token=False,
                                predictions=questions,
                                batch_size=16)
    return results

def get_lamma_perplexity(questions):
    perplexity = load("perplexity", module_type="metric")
    results = perplexity.compute(model_id='TheBloke/Llama-2-13B-GPTQ',
                                add_start_token=False,
                                predictions=questions,
                                batch_size=16)
    return results

```

```python
df = pd.read_csv('../test/genai_questions_809156896092057937.csv')
df = pd.read_csv('../test/genai_questions_311016922297059474.csv')
questions = df['questions'].to_list()

ans = get_perplexity(questions)
print (ans)
print (ans['mean_perplexity'])
```

```python
df = pd.read_csv('./existing_qa_datasets/squad_v2_q.csv')
questions = df['question'].to_list()

all_ans = []

batch_size = 400
for i in range(0, len(questions), batch_size):
    batch = questions[i:i + batch_size]
    batch_ans = get_perplexity(batch)
    all_ans.extend(batch_ans['perplexities'])

print (np.mean(all_ans))
```
