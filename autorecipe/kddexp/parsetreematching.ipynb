```python
import spacy
from spacy.matcher import Matcher

# Load the spaCy English model
nlp = spacy.load("en_core_web_sm")

# Define a spaCy Matcher
matcher = Matcher(nlp.vocab)

# Define the pattern for the template
pattern = [
    {"lower": "is"},                  # Case-insensitive match for "is"
    {"POS": {"in": ["DET", "NOUN"]}, "OP": "?"},  # Optional determiner or noun
    {"POS": "VERB"},                   # Verb
    {"POS": {"in": ["DET", "NOUN"]}, "OP": "?"},  # Optional determiner or noun
    {"IS_PUNCT": True}                 # Question mark
]

matcher.add("template", [pattern])

# Function to check if a question matches the template
def matches_template(question):
    doc = nlp(question)
    print (doc)
    matches = matcher(doc)
    print (matches)
    return bool(matches)

# Example usage
question = "is sky blue?"
if matches_template(question):
    print("The question matches the template.")
else:
    print("The question does not match the template.")
```
