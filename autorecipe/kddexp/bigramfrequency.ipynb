```python
import pandas as pd
df = pd.read_csv('../test/genai_questions_809156896092057937.csv')
print (df.head(10))
```

```python
questions = df['questions'].to_list()
```

```python
from nltk import bigrams
from collections import Counter
import nltk
nltk.download('punkt')  # Ensure that NLTK's punkt tokenizer is downloaded
```

```python
# Function to extract bigrams from the first two words of a sentence
def extract_bigrams(sentence):
    words = nltk.word_tokenize(sentence.lower())
    first_two_words = words[:2]
    return list(bigrams(first_two_words))

# Process each question and collect bigrams
all_bigrams = []
for question in questions:
    question_bigrams = extract_bigrams(question)
    all_bigrams.extend(question_bigrams)

# Get the top 10 most common bigrams
top_bigrams = Counter(all_bigrams).most_common(10)
```

```python
for bigram, count in top_bigrams:
    print(f"{bigram}: {count/len(questions)*100}")
```
