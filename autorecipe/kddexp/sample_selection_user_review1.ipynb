```python
import pandas as pd

ds_df = pd.read_csv("../../results/windturbinegearbox/genai_questions_answer_ds_bank_windturbinegearbox_llama.csv")
ds_df = ds_df[ds_df['round'] < 2]
ds_df = ds_df[['questions']]
ds_df['ds'] = 1

sme_df = pd.read_csv("../../results/windturbinegearbox/genai_questions_answer_sme_bank_windturbinegearbox_llama.csv")
sme_df = sme_df[sme_df['round'] < 2]
sme_df = sme_df[['questions']]
sme_df['sme'] = 1
```

```python
ds_df
```

```python
ans = ds_df.merge(sme_df, on='questions', how='outer')

ans.fillna(0, inplace=True)
ans['sum'] = ans['ds'] + ans['sme']
part1 = ans[ans['sum'] == 2]
part2 = ans[ans['sme'] == 1]
part3 = ans[ans['ds'] == 1]

both = part1['questions'].sample(7).to_list()
sme = part2['questions'].sample(7).to_list()
ds = part3['questions'].sample(6).to_list()

ansk = []
for item in both:
    ansk.append((item, 'Both DS & SME', 'Subject Matter Expert', 'Data Scientist', 'Both DS & SME', 'None of the above'))
for item in sme:
    ansk.append((item, 'Subject Matter Expert', 'Subject Matter Expert', 'Data Scientist', 'Both DS & SME', 'None of the above'))
for item in ds:
    ansk.append((item, 'Data Scientist', 'Subject Matter Expert', 'Data Scientist', 'Both DS & SME', 'None of the above'))

A = pd.DataFrame(ansk, columns=['Question', 'Real Persona', 'Persona Option 1', 'Persona Option 2', 'Persona Option 3', 'Persona Option 4'])
A.to_csv('set1.csv',index=False)
```
