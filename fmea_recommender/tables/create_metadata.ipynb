```python
import pandas as pd
import json
```

```python
dfs = []
```

```python
for i in range(10):
    df = pd.read_excel('all_iso.xlsx', sheet_name=i, header=1)
    dfs.append(df)
```

```python
metadata = {}
```

```python
for i, df in enumerate(dfs):
    if i == 1:
        continue
    for j in range(len(df)):
        row = df.iloc[j]
        cols = df.columns
        failure_code = row[cols[0]]
        desc = row[cols[1]]
        examples = row[cols[2]]
        components = []
        for col in df.columns[3:]:
            if row[col] == 'x':
                components.append(col)
        if not failure_code in metadata:
            metadata[failure_code] = {
                'description': desc,
                'examples': examples,
                'components': components
            }
        else:
            metadata[failure_code]['components'].extend(components)
```

```python
with open('metadata.json', 'w') as f:
    json.dump(metadata, f)
```
