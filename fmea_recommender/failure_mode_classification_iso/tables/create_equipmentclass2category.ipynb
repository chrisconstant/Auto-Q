```python
import pandas as pd
import json
```

```python
xl = pd.ExcelFile('EquiptmentCategory2Class.xlsx')
assets = xl.sheet_names  # see all sheet names
dfs = []
for asset in assets:
    df = pd.read_excel('EquiptmentCategory2Class.xlsx', sheet_name=asset)
    dfs.append(df)

df = df.ffill()
df = df[df.columns[:2]]
df = df.drop_duplicates()
df.head(100)
print (df.shape)

json_data = df.groupby(df.columns[0]).apply(lambda x: x.to_dict(orient='records')).to_json()

json_list = []
for group_name, group_data in df.groupby(df.columns[0]):
    group_values = group_data[df.columns[1]].tolist()  # Get values of the second column as a list
    json_list.append({group_name: group_values})

jsonlist = []
for it in json_list:
    for k in it:
        jsonlist.append({'Equipment Category': k, 'Equipment Class': it[k]})

print (jsonlist)

json_file_path = 'iso_equipment_category.json'

# Write the list of dictionaries to the JSON file
with open(json_file_path, 'w') as json_file:
    json.dump(jsonlist, json_file, indent=4)
```

```python
xl = pd.ExcelFile('equipment_class_type.xlsx')
assets = xl.sheet_names  # see all sheet names
dfs = []
for asset in assets:
    df = pd.read_excel('equipment_class_type.xlsx', sheet_name=asset)
    dfs.append(df)

df = df.ffill()
df = df[['Equipment class', 'Equipment type']]
df['Equipment class'] = df['Equipment class'].apply(lambda x: x if x[-2:] != ' a' else x[:-2])
df

json_data = df.groupby(df.columns[0]).apply(lambda x: x.to_dict(orient='records')).to_json()

json_list = []
for group_name, group_data in df.groupby(df.columns[0]):
    group_values = group_data[df.columns[1]].tolist()  # Get values of the second column as a list
    json_list.append({group_name: group_values})

jsonlist = []
for it in json_list:
    for k in it:
        jsonlist.append({'Equipment Class': k, 'Equipment Type': it[k]})

print (jsonlist)

json_file_path = 'iso_equipment_class_type.json'

# Write the list of dictionaries to the JSON file
with open(json_file_path, 'w') as json_file:
    json.dump(jsonlist, json_file, indent=4)
```

```python
xl = pd.ExcelFile('Equipment_unit_subunit_maintanable_component_revised.xlsx')
assets = xl.sheet_names  # see all sheet names
dfs = []
for asset in assets:
    df = pd.read_excel('Equipment_unit_subunit_maintanable_component_revised.xlsx', sheet_name=asset)
    dfs.append(df)
```

```python
iso_equipment_unit_items = []
for item in dfs:
    first_row = list(item.iloc[0][1:])
    updated_row = []
    for it_index, it in enumerate(first_row):
        try:
            if len(it) > 0:
                if it.endswith(' a'):
                    it = it[:-2]
                if it.endswith(' i'):
                    it = it[:-2]
                if it.endswith(' b'):
                    it = it[:-2]
                if it.endswith('/ '):
                    it = it[:-2]
                if it.endswith(' c'):
                    it = it[:-2]
                if it.endswith(' (at '):
                    it = it[:-5]
        except:
            pass

        cols = list(item.iloc[2:,it_index+1].dropna())
        updated_cols = []
        for it in cols:
            if len(it) > 0:
                if it.endswith(' a'):
                    it = it[:-2]
                if it.endswith(' b'):
                    it = it[:-2]
                if it.endswith('/ '):
                    it = it[:-2]
                if it.endswith(' c'):
                    it = it[:-2]
                if it.endswith(' e'):
                    it = it[:-2]
                if it.endswith(' (at '):
                    it = it[:-5]
                if it.endswith(' i'):
                    it = it[:-2]
            updated_cols.append(it)

        updated_row.append({'name':it, 'Maintainable items': updated_cols})

    a = ""
    iso_equipment_unit_items.append({'Equipment unit': item.columns[1], 'Subunit' : updated_row})

json_file_path = 'iso_equipment_unit_subunit.json'

# Write the list of dictionaries to the JSON file
with open(json_file_path, 'w') as json_file:
    json.dump(iso_equipment_unit_items, json_file, indent=4)
```
