```python
import json
```

```python
with open('metadata.json') as f:
    data = json.load(f)
```

```python
results = []
for item in data:    
    example_keys = ['workorderdescription', 'assetdescription', 'locationdescription', 'workordertaskdescription']
    built_str = f'Here is an example failure mode: {item["longdescription"].strip()}'
    results.append(built_str)
    for example in item['examples']:
        test = (
            f'A {example["persongroup"]} technician performed a work on the asset:{example["assetdescription"]} on the location {example["locationdescription"]} '
            f'and wrote: {example["workordertaskdescription"]}. This is a failure and is described as: {item["longdescription"]}'
        )
        built_str = f'Here is an example asset description: {example["assetdescription"].strip()}: '
        results.append(built_str)
        built_str = f'Here is an example work order description: {example["workorderdescription"].strip()}' 
        results.append(built_str)
        built_str = f'Here is an example location description: {example["locationdescription"].strip()}'
        results.append(built_str)
        built_str = f'Here is an example work order task description: {example["workordertaskdescription"].strip()}'
        results.append(built_str)
```
